Market Entry Analysisfor ABG Motors


```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets (clean extra columns)
in_df = pd.read_csv("C:/Users/user/Desktop/Market Analysis/IN_Data.csv", usecols=[0,1,2,3,4])
jpn_df = pd.read_csv("C:/Users/user/Desktop/Market Analysis/JPN_Data.csv", usecols=[0,1,2,3,4,5])

# Clean ANN_INCOME (remove commas/spaces)
for df in [in_df, jpn_df]:
    df["ANN_INCOME"] = pd.to_numeric(
        df["ANN_INCOME"].astype(str).str.replace(r"[\s,]", "", regex=True).str.strip(), 
        errors="coerce"
    )

print("Data Shapes:", in_df.shape, jpn_df.shape)
print(jpn_df["PURCHASE"].value_counts(normalize=True))  # ~57.6% purchase rate in Japan
```

```python
# Feature Engineering
jpn_df["GENDER_ENC"] = jpn_df["GENDER"].map({"M": 1, "F": 0})
in_df["GENDER_ENC"] = in_df["GENDER"].map({"M": 1, "F": 0})

# Additional engineered features
for df in [jpn_df, in_df]:
    df["AGE_INCOME_RATIO"] = df["CURR_AGE"] / (df["ANN_INCOME"] + 1)
    df["HIGH_INCOME"] = (df["ANN_INCOME"] > df["ANN_INCOME"].median()).astype(int)

# Impute AGE_CAR for India (market similarity assumption)
in_df["AGE_CAR"] = jpn_df["AGE_CAR"].median()
```

Classification Model on Japanese Dataset

```python
# Prepare features for model
features = ["CURR_AGE", "GENDER_ENC", "ANN_INCOME", "AGE_CAR"]  # Can add engineered ones
X = sm.add_constant(jpn_df[features])
y = jpn_df["PURCHASE"]

# Train model
model = sm.Logit(y, X).fit(disp=0)
print(model.summary())
```

Apply Model to Indian Dataset & Count Potential Customers

```python
# After training the model on JPN data...

# Prepare Indian data (ensure clean numeric columns)
in_df["GENDER_ENC"] = in_df["GENDER"].map({"M": 1, "F": 0})
in_df["AGE_CAR"] = jpn_df["AGE_CAR"].median()   # or any imputation

features = ["CURR_AGE", "GENDER_ENC", "ANN_INCOME", "AGE_CAR"]

# CRITICAL: Create X_in cleanly
X_in = in_df[features].copy()                    # explicit copy
X_in = sm.add_constant(X_in, has_constant='add')  # Force constant addition

# Predict
in_df["PRED_PROB"] = model.predict(X_in)
in_df["PREDICTED_PURCHASE"] = (in_df["PRED_PROB"] > 0.5).astype(int)

potential = in_df["PREDICTED_PURCHASE"].sum()
print(f"Potential customers: {potential} out of {len(in_df)}")
print(f"Percentage: {round(potential / len(in_df) * 100, 2)}%")
```

```python
# ====================== NEW FEATURES ======================

# 1. Interaction terms (very important for business insight)
jpn_df["AGE_CAR_x_INCOME"] = jpn_df["AGE_CAR"] * jpn_df["ANN_INCOME"]
in_df["AGE_CAR_x_INCOME"]   = in_df["AGE_CAR"]   * in_df["ANN_INCOME"]

# 2. Age Groups
for df in [jpn_df, in_df]:
    df["AGE_GROUP"] = pd.cut(df["CURR_AGE"], 
                             bins=[0, 30, 45, 60, 100], 
                             labels=["Young", "Middle", "Senior", "Elder"])

# 3. Income Tiers (more granular)
for df in [jpn_df, in_df]:
    df["INCOME_TIER"] = pd.qcut(df["ANN_INCOME"], q=5, labels=["Very Low", "Low", "Medium", "High", "Very High"])

# 4. Affordability Score
for df in [jpn_df, in_df]:
    df["AFFORDABILITY"] = df["ANN_INCOME"] / (df["CURR_AGE"] + 1)

# 5. Car Age Category
jpn_df["OLD_CAR"] = (jpn_df["AGE_CAR"] > jpn_df["AGE_CAR"].quantile(0.75)).astype(int)
in_df["OLD_CAR"]   = (in_df["AGE_CAR"]   > jpn_df["AGE_CAR"].quantile(0.75)).astype(int)
```

```python
# Updated feature list
features = ["CURR_AGE", "GENDER_ENC", "ANN_INCOME", "AGE_CAR", 
            "AGE_CAR_x_INCOME", "AFFORDABILITY", "OLD_CAR"]

# Retrain model
X = sm.add_constant(jpn_df[features])
model = sm.Logit(jpn_df["PURCHASE"], X).fit(disp=0)
print(model.summary())   # Check new coefficients
```

```python
# ====================== ADDITIONAL POWERFUL FEATURES ======================

# 6. Log Transformations (for skewed income)
for df in [jpn_df, in_df]:
    df["LOG_INCOME"] = np.log1p(df["ANN_INCOME"])

# 7. Age * Gender Interaction
for df in [jpn_df, in_df]:
    df["AGE_GENDER_INT"] = df["CURR_AGE"] * df["GENDER_ENC"]

# 8. Purchase Likelihood Score (simple composite)
jpn_df["PURCHASE_SCORE"] = (jpn_df["OLD_CAR"] * 0.4 + 
                            (jpn_df["HIGH_INCOME"] * 0.3) + 
                            (jpn_df["AGE_CAR"] / 1000 * 0.3))
```

```python
from sklearn.metrics import roc_auc_score, roc_curve, classification_report, confusion_matrix
import matplotlib.pyplot as plt

# Updated Features
features = ["CURR_AGE", "GENDER_ENC", "ANN_INCOME", "AGE_CAR", 
            "AGE_CAR_x_INCOME", "AFFORDABILITY", "OLD_CAR", 
            "LOG_INCOME", "AGE_GENDER_INT"]

# Train-Test Split for Proper Evaluation
from sklearn.model_selection import train_test_split
X = jpn_df[features]
y = jpn_df["PURCHASE"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

X_train_const = sm.add_constant(X_train)
X_test_const = sm.add_constant(X_test)

model = sm.Logit(y_train, X_train_const).fit(disp=0)
print(model.summary())

# Predictions
y_pred_prob = model.predict(X_test_const)
y_pred = (y_pred_prob > 0.5).astype(int)

# ====================== KEY METRICS ======================
print("=== MODEL PERFORMANCE ===")
print(classification_report(y_test, y_pred))
print("ROC AUC Score:", roc_auc_score(y_test, y_pred_prob))

# Confusion Matrix
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
plt.figure(figsize=(8,6))
plt.plot(fpr, tpr, label=f'ROC Curve (AUC = {roc_auc_score(y_test, y_pred_prob):.3f})')
plt.plot([0,1], [0,1], 'k--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve - Purchase Prediction')
plt.legend()
plt.savefig("roc_curve.png")
plt.show()
```
